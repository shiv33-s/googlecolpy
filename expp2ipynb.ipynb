#Here to implement simple linear regression

#Step 1:Importing Required Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

#Step 2:Read the Data


```python
data={"hours":[1,2,3,4,5],
      "marks":[50,60,80,90,100]}
df=pd.DataFrame(data)
print(df)

X=df[["hours"]].values
Y=df[["marks"]].values
```

#Step 3: Model Fitting

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X,Y)

slope=model.coef_[0]
intercept=model.intercept_
```

```python
#f-string (formatted string literal )in python
#2f this for mat specifies float within curly braces of an f-string
print(f"slope(m): {slope}")
print(f"intercept (b): {intercept}")
print(f"Equation of regression line: y={slope[0]:.2f} x + {intercept[0]:.2f}")
```

```python
df["Predicted Marks"]=model.predict(X)
```

```python
#here keep in mind the data which we have provided to the model is also linear
df.head()
```

#Step 5:Visualizing the Data

```python
plt.scatter(df['hours'],df['marks'],color='blue',label='Actual marks')
```

```python
# Plotting the equation
plt.plot(df['hours'],df['Predicted Marks'],color='red',label="Regression Line")
plt.xlabel('Hours Studied')
plt.ylabel('Marks')
plt.title('Linear Regression:hours vs marks')
plt.legend()
plt.show()
```
